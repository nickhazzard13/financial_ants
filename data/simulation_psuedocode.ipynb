Model 0 Psuedocode

```text
Level 0: One-Trader Momentum Model

SET starting capital to $10,000
SET evaluation period to 2024

LOAD SPY opening and closing prices
INCLUDE the final two trading days of 2023 for the first signal

INITIALIZE trader:
    Cash = $10,000
    Shares = 0

INITIALIZE separate buy-and-hold benchmark:
    Buy $10,000 of SPY at the first evaluation day's open
    Hold those shares throughout 2024

FOR each trading day in 2024:

    Change = previous close - the close before that
    Action = HOLD

    IF change > 0 AND cash > 0:
        Shares = cash / today's opening price
        Cash = 0
        Action = BUY

    ELSE IF change < 0 AND shares > 0:
        Cash = shares × today's opening price
        Shares = 0
        Action = SELL

    OTHERWISE:
        Keep current cash and shares

    AT today's close:
        Trader value = cash + shares × closing price
        Firm value = trader value

        Trader return (%) =
            100 × (trader value - starting capital) / starting capital

        Firm return (%) = trader return (%)

        

    RECORD:
        Date, opening price, closing price, signal, action
        Cash, shares, portfolio values, cumulative returns

DISPLAY final values and returns
PLOT trader/firm returns and benchmark returns
SAVE daily results
```

2 Trader, 2 Strategy Psuedocode


```text
Level 1: Two Traders, Two Strategies, Weekly Switching

SET starting firm capital to $10,000
SET starting capital per trader to $5,000
SET evaluation period to 2024

LOAD SPY opening and closing prices
INCLUDE the final two trading days of 2023 for the first signal

INITIALIZE Trader 1:
    Strategy = Momentum
    Cash = $5,000
    Shares = 0

INITIALIZE Trader 2:
    Strategy = Contrarian
    Cash = $5,000
    Shares = 0

FOR each trading day in 2024:

    IF this is the first trading day of a new calendar week:

        IF a previous evaluation week exists:

            FOR each strategy:
                Weekly return (%) =
                    100 × (scoring account's final value - 5,000) / 5,000

            IF momentum has the higher weekly return:
                Both traders select Momentum

            ELSE IF contrarian has the higher weekly return:
                Both traders select Contrarian

            OTHERWISE:
                Each trader keeps their current strategy

            Keep each trader's existing cash and shares

            RECORD:
                Review date, previous week's start and end dates
                Both strategy returns and each trader's selection

        INITIALIZE separate scoring accounts for the new week:
            Momentum account: $5,000 cash and zero shares
            Contrarian account: $5,000 cash and zero shares
            These hypothetical accounts add no money to the firm

    Change = previous close - the close before that

    IF change > 0:
        Momentum signal = BUY
        Contrarian signal = SELL

    ELSE IF change < 0:
        Momentum signal = SELL
        Contrarian signal = BUY

    OTHERWISE:
        Both strategy signals = HOLD

    AT today's open:

        FOR each trader and each strategy scoring account:

            Use the signal for its strategy
            Action = HOLD

            IF signal = BUY AND cash > 0:
                Shares = cash / today's opening price
                Cash = 0
                Action = BUY

            ELSE IF signal = SELL AND shares > 0:
                Cash = shares × today's opening price
                Shares = 0
                Action = SELL

            OTHERWISE:
                Keep current cash and shares

    AT today's close:

        FOR each trader:
            Trader value = cash + shares × closing price

            Trader return (%) =
                100 × (trader value - 5,000) / 5,000

        Firm value = Trader 1 value + Trader 2 value

        Firm return (%) =
            100 × (firm value - 10,000) / 10,000

        FOR each strategy scoring account:
            Account value = cash + shares × closing price
            Keep this value available for the next weekly review

    RECORD for each trader:
        Date, strategy, opening price, closing price, signal, action
        Cash, shares, trader value, trader return, firm value, firm return

DISPLAY final trader and firm values and returns
DISPLAY weekly strategy scores and selections
PLOT firm returns and SPY prices with buy and sell markers
PLOT Trader 1 and Trader 2 returns
SAVE daily results and weekly reviews
```
